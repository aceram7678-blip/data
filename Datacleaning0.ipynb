```r
normalise_zone <- function(x) {
  x <- trimws(toupper(x))
  x <- gsub("^CTR$", "CENTRAL", x)
  x <- gsub("^NORTH$", "NORTH", x)
  x <- gsub("^RIVERSIDE$", "RIVERSIDE", x)
  return(x)
}
customers$home_zone    <- normalise_zone(customers$home_zone)
drivers$base_zone      <- normalise_zone(drivers$base_zone)
vehicles$assigned_zone <- normalise_zone(vehicles$assigned_zone)
orders$pickup_zone     <- normalise_zone(orders$pickup_zone)
orders$dropoff_zone    <- normalise_zone(orders$dropoff_zone)
cat("Zone normalisation complete.\n")
```

```r
original_dispatch_time <- deliveries$dispatch_time
original_delivery_completed_at <- deliveries$delivery_completed_at

deliveries$dispatch_time <- ymd_hms(deliveries$dispatch_time)
deliveries$delivery_completed_at <- ymd_hms(deliveries$delivery_completed_at)

# Identify rows where parsing failed
failed_parse_rows <- which(is.na(deliveries$dispatch_time) | is.na(deliveries$delivery_completed_at))

if (length(failed_parse_rows) > 0) {
  cat(sprintf("Warning: %d rows failed to parse during datetime conversion.\n", length(failed_parse_rows)))
  cat("Original values for failed rows:\n")
  print(data.frame(
    delivery_id = deliveries$delivery_id[failed_parse_rows],
    original_dispatch_time = original_dispatch_time[failed_parse_rows],
    parsed_dispatch_time = deliveries$dispatch_time[failed_parse_rows],
    original_delivery_completed_at = original_delivery_completed_at[failed_parse_rows],
    parsed_delivery_completed_at = deliveries$delivery_completed_at[failed_parse_rows]
  ))
}

deliveries$actual_duration_hours = as.numeric(difftime(deliveries$delivery_completed_at, deliveries$dispatch_time, units = "hours"))

cat("Calculated actual_duration_hours and converted time columns to datetime.")
```
