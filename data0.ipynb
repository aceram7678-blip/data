## Removing Duplicate Rows

I will now check for and remove any duplicate rows from each DataFrame to ensure data integrity. I will display the number of duplicates before and after removal for each DataFrame.

```python
# Function to remove duplicates and print info
def remove_duplicates(df_name, df):
    initial_rows = df.shape[0]
    duplicates = df.duplicated().sum()
    if duplicates > 0:
        df = df.drop_duplicates().reset_index(drop=True)
        print(f"--- {df_name} DataFrame ---")
        print(f"Initial rows: {initial_rows}")
        print(f"Duplicate rows found and removed: {duplicates}")
        print(f"Rows after removing duplicates: {df.shape[0]}\n")
    else:
        print(f"--- {df_name} DataFrame ---")
        print(f"No duplicate rows found.")
        print(f"Total rows: {initial_rows}\n")
    return df

orders = remove_duplicates('orders', orders)
deliveries = remove_duplicates('deliveries', deliveries)
complaints = remove_duplicates('complaints', complaints)
app_events = remove_duplicates('app_events', app_events)
customers = remove_duplicates('customers', customers)
drivers = remove_duplicates('drivers', drivers)
hubs = remove_duplicates('hubs', hubs)
incidents = remove_duplicates('incidents', incidents)
vehicles = remove_duplicates('vehicles', vehicles)
data_dictionary = remove_duplicates('data_dictionary', data_dictionary)
```

## Analyzing Delivery Duration

I will calculate the delivery duration from the `deliveries` DataFrame by subtracting `dispatch_time` from `delivery_completed_at`. Then, I will convert this duration to hours and perform some basic descriptive statistics and visualization.

```python
# Calculate delivery duration
deliveries['delivery_duration'] = (deliveries['delivery_completed_at'] - deliveries['dispatch_time']).dt.total_seconds() / 3600

print("Descriptive statistics for Delivery Duration (hours):")
display(deliveries['delivery_duration'].describe())

# Visualize the distribution of delivery duration
plt.figure(figsize=(10, 6))
sns.histplot(deliveries['delivery_duration'], bins=30, kde=True)
plt.title('Distribution of Delivery Duration (Hours)')
plt.xlabel('Delivery Duration (Hours)')
plt.ylabel('Frequency')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
